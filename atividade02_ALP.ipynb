Questão #04 - estacionamento

```python
horas = int(input("Digite as horas:"))
if horas <= 2:
 valor = 8
elif horas <= 5:
 valor = 8 + 3 * (horas - 2)
else:
 valor = 17 + 2 * (horas - 5)

print("valor a pagar R$" , valor)
```

Questão #08

```python
anterior = int(input("número: "))
tem_iguais = False

for i in range(7):
  atual = int(input("Número: "))
  if atual == anterior:
    tem_iguais = True
  anterior = atual

if tem_iguais:
  print("Há números iguais e consecutivos")
else:
  print("Não há números iguais e consecutivos")
```

Questão #09

```python
saldo = 500.0
for i in range(6):
    tipo = int(input("Tipo (1 = depósito, 2 = saque)"))
    valor = float(input("Valor: "))
    if tipo == 1:
     saldo += valor
    elif tipo == 2:
     if valor <= saldo:
      saldo -= valor
     else:
       print("Saldo insuficiente")
print("Saldo final:", saldo)
```

Questão #11

```python

n = int(input("Digite um número maior que 1: "))
primo = True
i = 2

while i < n:
  if n % i == 0:
    primo = False
  i += 1
if primo:
  print("É primo")
else:
  print("Não é primo")
```

Questão #12

```python

n = int(input("Numero: "))
quantidade = 0

for i in range(1, n + 1):
  if n % i == 0:
    print(i)
    quantidade += 1
print("Quantidade de divisores:", quantidade)
```

Questão #13

```python

a = int(input("Numero: "))
b = int(input("Numero: "))
quantidade = 0

for num in range(a, b + 1):
  if num > 1:
    primo = True
    i = 2
    while i < num:
      if num % i == 0:
        primo = False
      i += 1
    if primo:
      print(num)
      quantidade += 1
print("Quantidade de primos: ", quantidade)
```
